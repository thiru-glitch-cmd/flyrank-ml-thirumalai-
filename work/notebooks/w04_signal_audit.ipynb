# ML-06 — Signal Audit

Audit feature signals, missing values, and relationship to the target label.

## 1. Missing value analysis

Analyze which columns have missing values and check whether the missingness is systematic.

```python
import pandas as pd
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
missing = df.isnull().mean() * 100
print(missing[missing > 0].sort_values(ascending=False))
```

## 2. Client-level distribution

Analyze the client distributions to see if some clients dominate the dataset.

```python
client_counts = df['client_id'].value_counts()
print(f"Top 5 clients row counts:\n{client_counts.head(5)}")
print(f"Share of top client: {client_counts.iloc[0] / len(df):.1%}")
```
