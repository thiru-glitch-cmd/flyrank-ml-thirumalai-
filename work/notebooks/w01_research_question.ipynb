# ML-02 — Research Question and Provisional Lane

This notebook defines the research question, chosen lane, and provisional dataset analysis.

## 1. My lane (or freestyle) and why

**Chosen Lane:** Lane 2 — Refresh / Content Opportunity Scoring.

**Rationale:** I have selected the Content Refresh Opportunity Scoring lane because in modern search engine optimization, content naturally decays over time. Reranking and identifying which pages require updates is highly actionable. Rather than manually auditing thousands of pages, a machine learning model can ingest multiple exposure, engagement, and content properties to surface a prioritized queue of content that is at risk of decay and has high search demand.

```python
import pandas as pd
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
print(f"Total rows in starter dataset: {len(df):,}")
print(f"Number of unique clients: {df['client_id'].nunique()}")
```

## 2. The question: decision, action, cost of a wrong call

**Question:** Out of the thousands of existing pages, which ones should be prioritized for review and refresh to prevent traffic decay?

**Decision/Action:** An editor or content manager will review the ranked list and decide whether to update the article, expand it with new sections, rewrite its meta description, or merge it with another page.

**Cost of a wrong call:**
- *False Positive (ranking a page high that doesn't need refresh):* Wasted editorial hours updating content that was already performing fine or had no potential.
- *False Negative (missing a page that is actively decaying):* Permanent loss of organic search visibility and sessions to competitors, causing a loss in business value.

```python
# Let's check the positive class rate (base rate of decline)
df['is_declining'] = df['trend_direction'].str.lower().eq('down').astype(int)
base_rate = df['is_declining'].mean()
print(f"Decline base rate: {base_rate:.1%}")
```

## 3. Quick look at the data (2-3 real numbers)

Looking at the dataset, we can observe critical statistics that highlight the size of the search demand and the opportunity to optimize.

```python
print(f"Median 90d Impressions: {df['impressions_90d'].median():.1f}")
print(f"Median 90d Sessions: {df['sessions_90d'].median():.1f}")
print(f"Mean content age (days): {df['content_age_days'].mean():.1f}")
```

## 4. Careful words: what I can and can't claim

**What we CAN claim:** We are building an observational decision-support model to prioritize human editorial attention. The model identifies signals correlated with historical organic visibility drops and suggests pages that are statistical candidates for refresh.

**What we CANNOT claim:** We cannot claim that our model represents the actual ranking algorithm of Google Search, nor can we guarantee that updating a recommended page will causally lead to a recovery (since we are observing correlations, not running a randomized controlled experiment).
