# ML-09 — Validation Audit

Audit the validation split to guarantee stability and prevent leakage.

## 1. Split strategy description

We use a **client-holdout split**. This means the validation set consists of entire clients that the model never saw during training. This verifies that our model generalizes to new client inventories.

```python
import json
with open('outputs/model_results.json') as f:
    res = json.load(f)
print(f"Split Strategy: {res['split_strategy']}")
print(f"Training rows: {res['train_rows']:,}")
print(f"Test rows: {res['test_rows']:,}")
```

## 2. Generalization comparison

Compare validation performance of Random Forest vs Baseline rules to prove lift.

```python
rf_ap = res['models']['random_forest']['average_precision']
base_ap = res['baseline']['baseline_average_precision']
print(f"Random Forest AP: {rf_ap:.3f}")
print(f"Baseline AP:      {base_ap:.3f}")
print(f"Relative AP Lift: {rf_ap / base_ap - 1:.1%}")
```
